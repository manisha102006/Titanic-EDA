```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("data/titanic.csv")
df.head()
```

# Chart 1: Overall Survival Distribution

```python
sns.countplot(data=df, x='Survived')
plt.title('Overall Survival Distribution')
plt.xlabel('Survival Status')
plt.ylabel('Passenger Count')
plt.xticks([0,1], ['Did Not Survive', 'Survived'])
```

Most passengers aboard the Titanic did not survive the disaster.

# Chart 2: Survival by Gender

```python
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Passenger Count')
```

Female passengers had substantially higher survival rates than male passengers

# Chart 3: Survival by Passenger Class

```python
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Passenger Count')
```

First-class passengers were more likely to survive than those in lower classes.

# Chart 4: Age Distribution by Survival

```python
sns.histplot(data=df, x='Age', hue='Survived',
             bins=30, kde=True, multiple='stack')
plt.title('Age Distribution by Survival')
plt.xlabel('Age')
plt.ylabel('Passenger Count')
```

Younger passengers showed slightly better survival outcomes than older passengers.

# Chart 5: Fare vs Survival

```python
sns.boxplot(data=df, x='Survived', y='Fare')
plt.title('Fare Distribution by Survival Status')
plt.xlabel('Survival Status')
plt.ylabel('Fare Paid')
plt.xticks([0,1], ['Did Not Survive', 'Survived'])
```

Survivors generally paid higher fares, suggesting an advantage associated with wealth and class.

# Chart 6: Embarkation Port vs Survival

```python
sns.countplot(data=df, x='Embarked', hue='Survived')
plt.title('Survival by Embarkation Port')
plt.xlabel('Embarkation Port')
plt.ylabel('Passenger Count')
```

Survival rates varied across embarkation ports, reflecting differences in passenger demographics.

The Titanic dataset reveals several clear patterns that influenced passenger survival. Overall, more passengers perished than survived, highlighting the severity of the disaster. Gender was one of the strongest predictors of survival, with women surviving at significantly higher rates than men, reflecting the “women and children first” evacuation practice.

Passenger class also played an important role. First-class passengers experienced much better survival outcomes than second- and third-class passengers, indicating that socioeconomic status may have affected access to lifeboats and evacuation routes. Age showed a weaker but noticeable relationship with survival, with younger passengers displaying somewhat higher survival rates.

Fare paid was another useful indicator. Survivors tended to have paid higher fares on average, which is closely related to passenger class and wealth. This suggests that wealthier passengers enjoyed advantages during evacuation. Finally, embarkation port displayed differences in survival outcomes, likely because each port had distinct passenger demographics and class distributions.

Taken together, the visualizations show that survival on the Titanic was not random. Factors such as gender, passenger class, age, fare, and embarkation location all influenced a passenger’s likelihood of survival. These charts demonstrate how exploratory data analysis can uncover meaningful stories hidden within historical datasets.
